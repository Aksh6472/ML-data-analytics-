```python
print("===== BEFORE =====")

print("Total nulls:", df.isnull().sum().sum())

print(df.isnull().sum())
```

```python
# Convert Timestamp into usable numeric features

if 'Timestamp' in df.columns:
    df['Timestamp'] = pd.to_datetime(
        df['Timestamp'],
        format='%d-%m-%Y %H:%M'
    )

    df['Hour'] = df['Timestamp'].dt.hour
    df['Month'] = df['Timestamp'].dt.month

    df = df.drop(columns=['Timestamp'])


# Numeric columns -> fill with median

numeric_cols = df.select_dtypes(
    include=['float64', 'int64']
).columns.tolist()

numeric_cols.remove('EnergyConsumption')

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())


# Categorical columns -> fill with mode

categorical_cols = df.select_dtypes(
    include=['object']
).columns.tolist()

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])


print("===== AFTER =====")

print("Total nulls:", df.isnull().sum().sum())

print(df.isnull().sum())
```

```python
# First, fill any potential NaNs in these columns with their mode before mapping
# Handle the case where a column might be entirely NaN, causing mode() to be empty.
default_fill_values = {
    'HVACUsage': 'Off',
    'LightingUsage': 'Off',
    'Holiday': 'No'
}

for col in ['HVACUsage', 'LightingUsage', 'Holiday']:
    if df[col].isnull().any(): # Only attempt to fill if there are NaNs
        modes = df[col].mode()
        if not modes.empty:
            df[col] = df[col].fillna(modes[0])
        else:
            # If mode is empty, it means all values in the column are NaN.
            # Fill with a predefined default value for this category type.
            df[col] = df[col].fillna(default_fill_values[col])

df['HVACUsage'] = df['HVACUsage'].map({
    'On': 1,
    'Off': 0
})

df['LightingUsage'] = df['LightingUsage'].map({
    'On': 1,
    'Off': 0
})

df['Holiday'] = df['Holiday'].map({
    'Yes': 1,
    'No': 0
})

# Check if 'DayOfWeek' exists before creating dummy variables
if 'DayOfWeek' in df.columns:
    df = pd.get_dummies(
        df,
        columns=['DayOfWeek'],
        drop_first=True
    )

df.head()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
X = df.drop(
    columns=['EnergyConsumption']
)

y = df['EnergyConsumption']
```

```python
scaler = StandardScaler()

X_scaled = pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)

print(
    "Train:",
    X_train.shape,
    "Test:",
    X_test.shape
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

single_feature_results = {}

for col in X_scaled.columns:

    Xc_train, Xc_test, yc_train, yc_test = train_test_split(
        X_scaled[[col]],
        y,
        test_size=0.2,
        random_state=42
    )

    m = LinearRegression().fit(
        Xc_train,
        yc_train
    )

    single_feature_results[col] = r2_score(
        yc_test,
        m.predict(Xc_test)
    )


# Rank features by how well they alone predict EnergyConsumption

ranked = pd.Series(
    single_feature_results
).sort_values(
    ascending=False
)

print(ranked)
```

```python
# Visualize the ranking

plt.figure(figsize=(10,6))

ranked.plot(
    kind='barh',
    color='teal'
)

plt.xlabel('R² Score (single feature)')

plt.title(
    'Which Single Feature Best Predicts EnergyConsumption?'
)

plt.gca().invert_yaxis()

plt.show()
```

```python
# Best single feature vs target - visualize the fit

best_feature = ranked.index[0]
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    X_scaled[best_feature],
    y,
    alpha=0.4,
    label='Actual'
)

m_best = LinearRegression().fit(
    X_scaled[[best_feature]],
    y
)
```

```python
plt.plot(
    X_scaled[best_feature],
    m_best.predict(
        X_scaled[[best_feature]]
    ),
    color='red',
    label='Regression Line'
)

plt.xlabel(best_feature)
plt.ylabel('EnergyConsumption')

plt.title(
    f'Best Fitting Feature: {best_feature}'
)

plt.legend()

plt.show()
```

```python
# Regression using HVACUsage

X_hvac = df[['HVACUsage']]
y = df['EnergyConsumption']

hvac_model = LinearRegression()
hvac_model.fit(X_hvac, y)

y_pred_hvac = hvac_model.predict(X_hvac)

plt.figure(figsize=(8, 5))

plt.scatter(
    df['HVACUsage'],
    y,
    alpha=0.5,
    label='Actual Data'
)

plt.plot(
    df['HVACUsage'],
    y_pred_hvac,
    linewidth=3,
    label='Regression Line'
)

plt.xlabel('HVAC Usage (0 = Off, 1 = On)')
plt.ylabel('Energy Consumption')
plt.title('Energy Consumption vs HVAC Usage')

plt.xticks([0, 1], ['OFF', 'ON'])

plt.legend()
plt.grid(True)

plt.show()
```

```python
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error
)

r2_hvac = r2_score(y, y_pred_hvac)

mae_hvac = mean_absolute_error(y, y_pred_hvac)

rmse_hvac = np.sqrt(
    mean_squared_error(y, y_pred_hvac)
)

mape_hvac = mean_absolute_percentage_error(
    y,
    y_pred_hvac
) * 100

print("===== HVAC REGRESSION RESULTS =====")
print(f"R² Score  : {r2_hvac:.4f}")
print(f"MAE       : {mae_hvac:.4f}")
print(f"RMSE      : {rmse_hvac:.4f}")
print(f"Error %   : {mape_hvac:.2f}%")
```

```python
model = LinearRegression()

model.fit(
    X_train,
    y_train
)

y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

r2 = r2_score(
    y_test,
    y_pred
)

mae = mean_absolute_error(
    y_test,
    y_pred
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)

print(f"R² Score : {r2:.4f}")
print(f"MAE : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
```

```python

# Coefficients - which features push predictions up/down the most

coeffs = pd.DataFrame({
    'Feature': X_scaled.columns,
    'Coefficient': model.coef_
}).sort_values(
    by='Coefficient',
    key=abs,
    ascending=False
)

coeffs
```
